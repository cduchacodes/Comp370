```python
#immport libraries
from datetime import datetime
import csv

start_date = datetime.strptime("01/01/2024", '%m/%d/%Y').date()
end_date = datetime.strptime("02/29/2024", '%m/%d/%Y').date() #leap year

```

```python
def most_complaint(start, end):
    counts = {}
    with open("hw5.csv", 'r') as fi:
        reader = csv.DictReader(fi)
    
        #go through each line
        for line in reader:
            created_date = datetime.strptime(line['Created Date'], '%m/%d/%Y %I:%M:%S %p').date()
    
            if created_date >= start and created_date <= end:
                key = line['Complaint Type']
                counts[key] = counts.get(key, 0) + 1
        return counts
```

```python
counts = most_complaint(start_date, end_date)
top_complaint = max(counts, key=counts.get)
print(f"Top complaint: {top_complaint} ({counts[top_complaint]} occurrences)")
```

```python
june_date = datetime.strptime("06/01/2024", '%m/%d/%Y').date()
july_date = datetime.strptime("07/31/2024", '%m/%d/%Y').date()
counts2 = most_complaint(june_date, july_date)
```

```python
#Make bar chart
import matplotlib.pyplot as plt

dates = ["January-February", "June-July"]
countss = [counts[top_complaint], counts2[top_complaint]]

bars = plt.bar(dates, countss)
```
